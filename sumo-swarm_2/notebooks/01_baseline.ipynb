```python
from env import SumoSwarmEnv
from config import ROUTE_FILES
import numpy as np, traci

env = SumoSwarmEnv(route_file=ROUTE_FILES['peak'])
obs = env.reset()
done = False
waits = []

while not done:
    # fixed-time: always keep current phase (action=0)
    actions = {aid: 0 for aid in env.agent_ids}
    obs, rewards, done, _ = env.step(actions)
    waits.append(-np.mean(list(rewards.values())))

env.close()

print(f'Fixed-time baseline — mean wait: {np.mean(waits):.2f}s')
print(f'This is the number all trained agents must beat.')
```
